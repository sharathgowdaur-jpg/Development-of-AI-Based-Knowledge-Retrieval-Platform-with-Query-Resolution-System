Filter employees whose salary is greater than 50000 using the PySpark filter() method.

Store the filtered results in a new DataFrame named high_salary_employees.

Sort all employee records by Salary in descending order, from highest to lowest.

Use the PySpark orderBy() method and desc() function.

```python
from pyspark.sql.functions import desc
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType


schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone", StringType(), True)
])

data = [
    ("Alice", "HR", 60000, "alice@example.com", "111-222-3333"),
    ("Bob", "Engineering", 90000, "bob@example.com", "444-555-6666"),
    ("Charlie", "HR", 65000, "charlie@example.com", "777-888-9999"),
    ("David", "Sales", 75000, "david@example.com", "123-456-7890"),
    ("Eve", "Engineering", 100000, "eve@example.com", "987-654-3210")
]

employees_df = spark.createDataFrame(data, schema)

print("Original Employees DataFrame:")
employees_df.show(truncate=False)

# Sort employee records by Salary in descending order
sorted_employees = employees_df.orderBy(desc("Salary"))

print("Employees DataFrame sorted by Salary (descending):")
# Display all sorted records
sorted_employees.show(truncate=False)

```

Reuse the existing SparkSession named spark and DataFrame named employees_df.

The DataFrame contains these columns: Name, Department, Salary, Email, and Phone.

Use the PySpark max() aggregation function to find the highest salary.

Store the result in a variable named highest_salary.

Display the highest salary with a clear message.

If the DataFrame is empty or all salary values are null, display a suitable message.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import max

# Reuse the existing SparkSession named spark
# Assume 'spark' is already initialized. If not, initialize it:
# spark = SparkSession.builder.appName("HighestSalary").getOrCreate()

# Assume 'employees_df' is an existing DataFrame.
# For demonstration purposes, let's create a sample employees_df.
# In a real scenario, this DataFrame would already be available.
data = [
    ("Alice", "HR", 60000, "alice@example.com", "555-1111"),
    ("Bob", "Engineering", 90000, "bob@example.com", "555-2222"),
    ("Charlie", "HR", 65000, "charlie@example.com", "555-3333"),
    ("David", "Sales", 75000, "david@example.com", "555-4444"),
    ("Eve", "Engineering", 95000, "eve@example.com", "555-5555"),
    ("Frank", "Sales", None, "frank@example.com", "555-6666"), # Example with null salary
    ("Grace", "Marketing", 70000, "grace@example.com", "555-7777")
]
columns = ["Name", "Department", "Salary", "Email", "Phone"]
employees_df = spark.createDataFrame(data, columns)

# --- Scenario 1: Empty DataFrame (for testing the empty check)
# employees_df = spark.createDataFrame([], columns)

# --- Scenario 2: All salaries are null (for testing the all-null check)
# data_all_null_salary = [
#     ("John", "HR", None, "john@example.com", "555-1000"),
#     ("Jane", "Engineering", None, "jane@example.com", "555-2000")
# ]
# employees_df = spark.createDataFrame(data_all_null_salary, columns)

# Check if the DataFrame is empty first
if employees_df.count() == 0:
    print("The DataFrame is empty. Cannot determine the highest salary.")
else:
    # Use the max() aggregation function to find the highest salary
    # .collect()[0][0] extracts the scalar value from the DataFrame row
    highest_salary = employees_df.agg(max("Salary")).collect()[0][0]

    # Display the result with a clear message, handling cases where all salaries are null
    if highest_salary is None:
        print("All salary values in the DataFrame are null. Cannot determine a highest non-null salary.")
    else:
        print(f"The highest salary is: {highest_salary}")

```

Reuse the existing SparkSession named spark and DataFrame named employees_df.

The DataFrame has these columns: Name, Department, Salary, Email, and Phone.

Use the PySpark groupBy() and count() methods to count employees in each department.

Group the records by the Department column.

Rename the count column to employee_count.

Sort the results by department name in ascending order.

Display the results using show(truncate=False).

```python
# Assuming 'spark' and 'employees_df' are already defined and available
# For demonstration purposes, let's create a sample employees_df if it doesn't exist
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# Initialize SparkSession if not already initialized (for standalone execution)
try:
    spark
except NameError:
    spark = SparkSession.builder.appName("EmployeeAnalysis").getOrCreate()

# Create a sample employees_df DataFrame for demonstration purposes
# In a real scenario, employees_df would already be loaded from a previous step.
if 'employees_df' not in locals():
    data = [
        ("Alice", "HR", 60000, "alice@example.com", "111-222-3333"),
        ("Bob", "Engineering", 80000, "bob@example.com", "444-555-6666"),
        ("Charlie", "HR", 65000, "charlie@example.com", "777-888-9999"),
        ("David", "Sales", 70000, "david@example.com", "000-111-2222"),
        ("Eve", "Engineering", 90000, "eve@example.com", "333-444-5555"),
        ("Frank", "Sales", 72000, "frank@example.com", "666-777-8888"),
        ("Grace", "HR", 62000, "grace@example.com", "999-000-1111"),
    ]
    columns = ["Name", "Department", "Salary", "Email", "Phone"]
    employees_df = spark.createDataFrame(data, columns)

# 1. Group the records by the Department column and count employees
# 2. Rename the count column to employee_count
# 3. Sort the results by department name in ascending order
department_employee_counts = employees_df.groupBy("Department") \
                                         .count() \
                                         .withColumnRenamed("count", "employee_count") \
                                         .orderBy(col("Department").asc())

# 4. Display the results using show(truncate=False)
print("Employee counts per department:")
department_employee_counts.show(truncate=False)

# Stop the SparkSession if it was started here and not needed further
# spark.stop()
```

Store the selected columns in a new DataFrame named selected_employees.

```python
# Assume 'spark' SparkSession and 'employees_df' DataFrame are already defined.
# For demonstration purposes, let's create a sample employees_df if it doesn't exist.
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Initialize SparkSession if not already available
try:
    spark
except NameError:
    spark = SparkSession.builder.appName("SelectColumns").getOrCreate()

# Define schema for the sample DataFrame
schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone", StringType(), True)
])

# Sample data
data = [
    ("Alice", "HR", 60000, "alice@example.com", "111-222-3333"),
    ("Bob", "Engineering", 90000, "bob@example.com", "444-555-6666"),
    ("Charlie", "HR", 65000, "charlie@example.com", "777-888-9999"),
    ("David", "Sales", 75000, "david@example.com", "000-111-2222"),
    ("Eve", "Engineering", 95000, "eve@example.com", "333-444-5555")
]

# Create the employees_df DataFrame
employees_df = spark.createDataFrame(data, schema=schema)

print("Original employees_df DataFrame:")
employees_df.printSchema()
employees_df.show(truncate=False)

# Select only the Name, Department, and Salary columns
selected_employees = employees_df.select("Name", "Department", "Salary")

print("\nDataFrame with selected columns (Name, Department, Salary):")
selected_employees.show(truncate=False)

# If you need to stop the SparkSession uncomment the line below
# spark.stop()
```
